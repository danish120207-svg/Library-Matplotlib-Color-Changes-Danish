# Semua bar menggunakan satu warna

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

plt.bar(nama, markah, color="pink")

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

# Setiap bar menggunakan warna berbeza

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna = ["pink", "red", "blue", "brown", "yellow"]

plt.bar(nama, markah, color=warna)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

# Pengguna memilih warna graf

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna_pilihan = input(
"Masukkan warna graf seperti pink, red, green atau blue: "
)

plt.bar(nama, markah, color=warna_pilihan)
plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

# Ubah warna bar, bingkai dan latar belakang

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

plt.bar(
nama,
markah,
color="pink",
edgecolor="black",
linewidth=2
)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.gca().set_facecolor("lightgreen")

plt.show()
```

# Ubah warna berdasarkan markah

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna = []

for nilai in markah:
    if nilai >= 80:
        warna.append("green")
    else:
        warna.append("red")
plt.bar(nama, markah, color=warna)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")
plt.show()
```
